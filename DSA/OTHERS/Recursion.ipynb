```python
# factorial of a number
def factorial(n):
    if n==0:
         return 1
    
    else:
        return n*factorial(n-1)

print(factorial(5))
```

```python
# sum of fibonaci series:
def fibo(m):
    if m==0:
        return 0
    elif m==1:
        return 1
    else:
         return fibo(m-1)+fibo(m-2)
print(fibo(10))
```
