```python
def subarraySum(nums, k):
    prefix_map = {0: 1}
    current_sum = 0
    count = 0

    for num in nums:
        current_sum += num
        count += prefix_map.get(current_sum - k, 0)
        prefix_map[current_sum] = prefix_map.get(current_sum, 0) + 1

    return count
```

```python
def first_unique(arr):
    freq = {}

    for x in arr:
        freq[x] = freq.get(x, 0) + 1

    for x in arr:
        if freq[x] == 1:
            return x

    return -1
```

```python
def frequency_map(arr):
    freq = {}

    for x in arr:
        freq[x] = freq.get(x, 0) + 1

    return freq
```
