```python
# implementing stack using the quene
class Stack:
    def __init__(self) -> None:
        self.dq1 = deque
        self.dq2 = deque
        
    def append(self,val):
        self.dq1.append(val)
        
        while self.dq1:
            self.dq2.append(self.dq1.popleft())
        
        self.dq1, self.dq2 = self.dq2, self.dq1

    def pop(self):
        if not self.dq1:
            return "No Elements"
        return self.dq1.popleft()

t = Stack()
```
