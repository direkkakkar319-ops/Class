```python
# reversing a linked list
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None

# declaring nodes
a = Node(1)
b = Node(2)
c = Node(3)
d = Node(4)
e = Node(5)
f = Node(6)

# connecting nodes
a.next = b
b.next = c
c.next = d
d.next = e
e.next = f

# declaring head of the linked list
head = a

# printing the current linked list
curr = head
print("original list")
while curr is not None:
    print(curr.data,end=" ")
    curr = curr.next

curr = head 
prev = None
nxt = None
while curr is not None:
    nxt = curr.next
    curr.next = prev
    prev = curr
    curr = nxt

head = prev
curr = head
print("\nreversed list")
while curr is not None:
    print(curr.data,end=" ")
    curr = curr.next
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, data):
        newNode = Node(data)
        
        if self.head is None:
            self.head = newNode
            return

        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = newNode

    def display(self):
        curr = self.head
        while curr:
            print(curr.data, end=" -> ")
            curr = curr.next
        print("None")
# ------------ MAIN CODE -------------
l = LinkedList()

n = int(input("enter the number of nodes you want: "))

for i in range(n):
    val = int(input(f"enter value {i+1}: "))
    l.insert(val)
    print("linked list:")
    l.display()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, data):
        newNode = Node(data)
        if self.head is None:
            self.head = newNode
            return 
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = newNode
    
    def display(self):
        curr = self.head
        while curr:
            print(curr.data,end="-->")
            curr = curr.next
        print("None")

l = LinkedList()
n=int(input("enter total number of nodes to be entered"))

for i in range(n):
    val=int(input(f"enter {i+1}node's value"))
    l.insert(val)
    l.display()

print("\nFinal Linked List is")
l.display()
```

```python
class ListNode:
    def __init__(self,val):
        self.val = val
        self.next = None
class LinkedList:
    def __init__(self):
        self.head = None
    def insertNode(self,val):
        newNode = ListNode(val)
        if self.head is None:
            self.head = newNode
            return
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = newNode
    def display(self):
        curr = self.head
        while curr:
            print(curr.val,end="-->")
            curr = curr.next
        print("None")
l = LinkedList()
n=int(input("enter number of nodes to be entered"))
for i in range(n):
    val = int(input(f"enter {i+1} node's value"))
    l.insertNode(val)
    l.display()

print("\nfinal nodes")
l.display()
```

```python
def findMiddle(head):
    slow = fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

    return slow
```
