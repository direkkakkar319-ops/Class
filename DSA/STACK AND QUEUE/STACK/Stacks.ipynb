## Array implementation of stack

```python
# creating a list
st = []

# appending elements(pushing)
st.append(1)
st.append(2)
st.append(3)
st.append(4)
st.append(5)

print(st)

# removing the last element
print(st.pop())

# accessing the last element from the array [will be at the top of the stack](top-function)
print(st[-1])
```

## Linked List Implementation of Stack  

```python
class ListNode:
    def __init__(self, val):
        self.data = val
        self.next = None
class Stack:
    def __init__(self):
        self.top = None
        self.length = 0

    def push(self, x):
        if self.top is None:
            self.top = ListNode(x)
            self.length += 1
            return

        newNode = ListNode(x)
        newNode.next = self.top
        self.top = newNode
        self.length += 1
    
    def pop(self):
        if self.top is None:
            return -1

        x = self.top.data
        self.top = self.top.next
        self.length -= 1
        return x
    
    def getTop(self):
        if self.top is None:
            return -2
        return self.top.data
    
    def size(self):
        return self.length
stack = Stack()

stack.push(1)
stack.push(2)
stack.push(3)

print(stack.size())
print(stack.pop())
print(stack.getTop())
```
